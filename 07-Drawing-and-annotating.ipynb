<div class="alert alert-warning">
From the documentation, **OpenCV changes its behavior according to the type** of the array:
    <ul>
    <li>If the image is 8-bit unsigned, it is displayed as is.</li>
    <li>If the image is 16-bit unsigned or 32-bit integer, the pixels are divided by 256. That is, the value range [0,255*256] is mapped to [0,255].</li>
    <li>If the image is 32-bit floating-point, the pixel values are multiplied by 255. That is, the value range [0,1] is mapped to [0,255].</li>
    </ul>

```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

# ساخت تصویر سیاه
image = np.zeros((512,512,3), np.uint8)

# رسم مستطیل: از (100,100) تا (300,250) با رنگ قرمز و ضخامت ۵
cv2.rectangle(image, (100,100), (300,250), (0,0,255), 5)

# نمایش تصویر با cv2_imshow
cv2_imshow(image)

# نمایش تصویر با matplotlib (تبدیل BGR به RGB)
plt.imshow(image[...,::-1])
plt.title("Rectangle")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

# ساخت تصویر سیاه
ellipse = np.zeros((512,512,3), np.uint8)

# رسم بیضی:
# مرکز: (250,250)
# شعاع‌ها: (250,100)
# زاویه دوران: 45 درجه
# از 90 تا 360 درجه
# رنگ: سفید
# ضخامت: -1 (پر شده)
cv2.ellipse(ellipse, (250, 250), (250, 100), 45, 90, 360, (255,255,255), -1)

# نمایش تصویر با cv2_imshow
cv2_imshow(ellipse)

# نمایش تصویر با matplotlib (تبدیل BGR به RGB)
plt.imshow(ellipse[...,::-1])
plt.title("Ellipse")
plt.axis('off')
plt.show()
```

## <div style="direction:rtl;text-align:right;font-family:B Lotus, B Nazanin, Tahoma">ترسیم چند ضلعی</div>


```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

# ساخت تصویر سیاه
image = np.zeros((512,512,3), np.uint8)

# تعریف چهار نقطه
pts = np.array([[10,50], [400,50], [90,200], [50,500]], np.int32)

# تبدیل نقاط به شکل مورد نیاز تابع polylines
pts = pts.reshape((-1,1,2))

# رسم چندضلعی
cv2.polylines(image, [pts], True, (0,0,255), 3)

# نمایش تصویر با cv2_imshow
cv2_imshow(image)

# نمایش تصویر با matplotlib (تبدیل BGR به RGB)
plt.imshow(image[...,::-1])
plt.title("Polygon")
plt.axis('off')
plt.show()
```

 <div class="alert alert-warning">
If third argument is False, you will get a polylines joining all the points, not a closed shape.<br>
**cv2.polylines()** can be used to draw multiple lines. Just create a list of all the lines you want to draw and pass it to the function.<br>
All lines will be drawn individually. It is a much better and faster way to draw a group of lines than calling **cv2.line()** for each line.
<p>https://docs.opencv.org/3.1.0/dc/da5/tutorial_py_drawing_functions.html</p>

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# ساخت تصویر سیاه
image = np.zeros((512,512,3), np.uint8)

# تعریف متن فارسی
text = "سلام دنیا"

# نوشتن متن روی تصویر
cv2.putText(image, text, (75,290), cv2.FONT_HERSHEY_COMPLEX, 2, (100,170,0), 3)

# نمایش تصویر با matplotlib (تبدیل BGR به RGB)
plt.imshow(image[...,::-1])
plt.title("سلام دنیا")
plt.axis('off')
plt.show()
```

```python
# ساخت تصویر سیاه
image = np.zeros((512,512,3), np.uint8)

# متن فارسی
text_to_be_reshaped = 'سلام دنیا'
reshaped_text = arabic_reshaper.reshape(text_to_be_reshaped)
bidi_text = get_display(reshaped_text)

# تبدیل تصویر OpenCV به PIL
pil_image = Image.fromarray(image)

# بارگذاری فونت
font = ImageFont.truetype("/content/Vazirmatn-Black.ttf", 60, encoding='unic')

# نوشتن متن
draw = ImageDraw.Draw(pil_image)

# محاسبه اندازه متن (با textbbox)
bbox = draw.textbbox((0,0), bidi_text, font=font)
text_width = bbox[2] - bbox[0]
text_height = bbox[3] - bbox[1]

# تعیین موقعیت برای راست‌چین
right_margin = 20
x_position = image.shape[1] - text_width - right_margin
y_position = 220

# نوشتن متن
draw.text((x_position, y_position), bidi_text, fill=(100,170,0), font=font)

# نمایش تصویر
plt.imshow(pil_image)
plt.axis('off')
plt.title("نوشتن متن فارسی راست‌چین")
plt.show()
```
